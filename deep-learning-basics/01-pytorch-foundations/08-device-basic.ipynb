```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. 사용 가능한 device 확인하기
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('선택된 device:', device)

x = torch.randn(3, 4)
# TODO: x를 device로 이동해보세요.
x_device = x.to(device)
print('x_device.device:', x_device.device)
```

```python
# 필수 2. 모델과 입력 Tensor를 같은 device로 이동하기
model = nn.Linear(4, 2)
x = torch.randn(5, 4)

# TODO: model과 x를 같은 device로 이동해보세요.
model = model.to(device)
x = x.to(device)
out = model(x)
print('out shape:', out.shape)

print('out shape :', out.shape)
print('out device:', out.device) # 실제 out이 위치한 장치 출력

assert out.device.type == device.type
print("성공적으로 장치가 일치합니다!")
```

```python
# 심화 1. 안전한 device 이동 함수 만들기
def move_batch_to_device(batch, device):
    x, y = batch
    x = x.to(device)
    y= y.to(device)
    # TODO: x와 y를 모두 device로 이동해서 반환하세요.
    return x, y

batch = (torch.randn(2, 4), torch.tensor([0, 1]))
x_moved, y_moved = move_batch_to_device(batch, device)
print(x_moved.device, y_moved.device)
```
