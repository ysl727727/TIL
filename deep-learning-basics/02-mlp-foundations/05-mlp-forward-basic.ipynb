```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. 간단한 MLP 클래스 뼈대 완성하기   ** 강의시간에 배운대로 __init__ 안에 relu를 선언하고 forward에 적용함
# 답안은 forward에 torch.relu로 따로 선언해서 적용.
class TinyMLP(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        # TODO: fc1 -> ReLU -> fc2 순서로 작성하세요.
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x

model = TinyMLP(4, 8, 3)
x = torch.randn(5, 4)
logits = model(x)
print('logits shape:', logits.shape)
```

```python
####

# 필수 1. 간단한 MLP 클래스 뼈대 완성하기
class TinyMLP(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        h = torch.relu(self.fc1(x))
        return self.fc2(h)

model = TinyMLP(4, 8, 3)
x = torch.randn(5, 4)
logits = model(x)
print('logits shape:', logits.shape)
assert logits.shape == (5, 3)
```

```python
# 필수 2. output_dim을 class 수와 맞추기
num_classes = 3
# TODO: 출력 차원이 num_classes와 같은지 확인해보세요.
print('num_classes:', num_classes)
print('model output dim:', logits.shape[-1])

assert logits.shape[-1] == num_classes
```

```python
# 심화 1. hidden_dim을 바꿔도 동작하는지 확인하기
for hidden_dim in [4, 16]:
    model = TinyMLP(input_dim=4, hidden_dim=hidden_dim, output_dim=3)
    logits = model(torch.randn(2, 4))
    print('hidden_dim:', hidden_dim, 'logits shape:', logits.shape)
    assert logits.shape == (2, 3)
```
