강의 원문 대신 직접 작성하고 실행한 코드와 학습 메모를 정리했습니다.


```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. ReLU 전후 값 비교
values = torch.tensor([-2.0, -0.5, 0.0, 1.0, 3.0])
# TODO: torch.relu를 적용해보세요.
relu_values = torch.relu(values)
print('before:', values.tolist())
print('after:', relu_values.tolist())

assert relu_values.tolist() == torch.tensor([0.0, 0.0, 0.0, 1.0, 3.0]).tolist()
```

```python
# 필수 2. MLP 은닉층에 ReLU 추가하기
model = nn.Sequential(
    nn.Linear(4, 8),
    # TODO: 여기에 ReLU를 추가해보세요.
    nn.ReLU(),
    nn.Linear(8, 3)
)
x = torch.randn(2, 4)
logits = model(x)
print('logits shape:', logits.shape)

assert logits.shape == torch.Size([2,3])
```

```python
# 심화 1. dead ReLU 직관 확인하기
negative_input = torch.tensor([-5.0, -1.0, 0.5])
print('ReLU output:', torch.relu(negative_input).tolist())
print('음수 입력은 0으로 잘리므로 일부 뉴런이 계속 0만 출력할 수 있습니다.')

```
