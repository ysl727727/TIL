```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. logits를 Sigmoid 확률로 바꾸기
logits = torch.tensor([-2.0, 0.0, 2.0])
# TODO: sigmoid를 적용해보세요.
probs = torch.sigmoid(logits)
print('logits:', logits.tolist())
print('probs:', probs.tolist())

assert torch.all((probs >= 0) & (probs <= 1))
```

```python
# 필수 2. threshold로 이진 label 만들기
logits = torch.tensor([-1.2, 0.4, 2.1, -0.7])
probs = torch.sigmoid(logits)
# TODO: 확률이 0.5 이상이면 1, 아니면 0으로 예측하세요.
pred_label = (probs >= 0.5).long() ## 참이면 1 거짓이면 0 각각 lnt64로 반환.
print('probs:', probs.round(decimals=3).tolist())
print('pred_label:', pred_label.tolist())

assert pred_label.tolist() == [0, 1, 1, 0]
```
