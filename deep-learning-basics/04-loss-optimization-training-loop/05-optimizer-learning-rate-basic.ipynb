# 5-3 Optimizer와 Learning Rate

강의 원문 대신 직접 작성한 코드, 실행 결과와 학습 메모를 정리했습니다.


```python
import random
import json
import math
import shutil
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader, Dataset, random_split

# 실습 결과가 매번 비슷하게 나오도록 seed를 고정합니다.
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('device:', device)
```

```python
w = torch.tensor([0.0], requires_grad=True)
x = torch.tensor([2.0])
y = torch.tensor([6.0])

# TODO: SGD optimizer를 생성하세요. lr은 0.1로 설정합니다.
optimizer = torch.optim.SGD([w], lr=0.1)  # 임시 코드입니다.

pred = w * x
loss = (pred - y).pow(2).mean()

# TODO: gradient 초기화, backward, step 순서로 작성하세요.
optimizer.zero_grad()
loss.backward()
optimizer.step()

print('loss:', float(loss))
print('updated w:', float(w.detach()))
```

```python
def run_lr(lr, epochs=20):
    torch.manual_seed(0)
    model = nn.Linear(1, 1)
    x = torch.linspace(-1, 1, 40).view(-1, 1)
    y = 2 * x + 1
    optimizer = torch.optim.SGD(model.parameters(), lr=lr)
    loss_fn = nn.MSELoss()
    history = []
    for epoch in range(epochs):
        pred = model(x)
        loss = loss_fn(pred, y)
        # TODO: optimizer step 순서를 완성하세요.
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        history.append(float(loss.detach()))
    return history

# TODO: 작은 lr, 적당한 lr, 큰 lr을 넣어 비교하세요.
lrs = [0.0001, 0.01, 0.5]
for lr in lrs:
    h = run_lr(lr)
    print('lr:', lr, 'first:', round(h[0], 4), 'last:', round(h[-1], 4))
```

```python
def make_optimizer(name, params, lr):
    # TODO: name에 따라 optimizer를 반환하세요.
    if name == 'sgd':
        return torch.optim.SGD(params, lr=lr)  # 임시 코드입니다.
    if name == 'adam':
        return torch.optim.Adam(params, lr=lr)  # 임시 코드입니다.
    raise ValueError(name)

model = nn.Linear(2, 1)
for name in ['sgd', 'adam']:
    opt = make_optimizer(name, model.parameters(), lr=0.01)
    print(name, '->', opt.__class__.__name__)
```
