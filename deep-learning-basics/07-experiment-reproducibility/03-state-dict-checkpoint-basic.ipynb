# [9-3강] state_dict와 checkpoint 구성 - 실습

```python
import random
import json
import math
import shutil
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader, Dataset, random_split

# 실습 결과가 매번 비슷하게 나오도록 seed를 고정합니다.
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('device:', device)
```

## 필수 1 1: state_dict key 확인하기

### 문제 설명
모델의 state_dict에서 layer별 weight/bias key를 확인합니다.

```python
model = nn.Sequential(nn.Linear(3, 4), nn.ReLU(), nn.Linear(4, 2))
# TODO: state_dict를 가져오세요.
state = model.state_dict()
for k, v in state.items():
    print(k, tuple(v.shape))
```

### 해설 및 실행 결과 해석
state_dict에는 모델 구조 자체가 아니라 layer별 Tensor 값이 저장됩니다. 같은 구조의 모델을 만든 뒤 이 값을 불러와야 합니다.

## 필수 2 2: checkpoint dict 만들기

### 문제 설명
모델, optimizer, epoch, config를 하나의 dict로 묶습니다.

```python
model = nn.Linear(2, 1)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
config = {'lr': 0.01, 'model': 'linear'}
# TODO: checkpoint dict를 완성하세요.
checkpoint = {
    'epoch': 5,
    'model_state': model.state_dict(),
    'optimizer_state': optimizer.state_dict(),
    'config': config
}
print(checkpoint.keys())
```

### 해설 및 실행 결과 해석
checkpoint는 학습을 이어가거나 결과를 재현하기 위한 묶음입니다. model_state만 저장하면 추론은 가능하지만 학습 재개 정보는 부족합니다.
