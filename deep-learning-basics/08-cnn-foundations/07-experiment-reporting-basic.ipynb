## 문제 2. best epoch와 train-valid loss gap 계산하기

epoch별 기록에서 validation loss가 가장 낮은 epoch를 찾고, 그 epoch의 train-valid loss gap을 계산합니다.


```python
history = [
    {'epoch': 1, 'train_loss': 0.70, 'valid_loss': 0.68, 'valid_acc': 0.50},
    {'epoch': 2, 'train_loss': 0.42, 'valid_loss': 0.40, 'valid_acc': 0.88},
    {'epoch': 3, 'train_loss': 0.18, 'valid_loss': 0.29, 'valid_acc': 0.90},
]
# TODO: valid_loss 기준 best epoch를 찾고 loss_gap을 계산하세요.
best = min(history, key=lambda x: x['valid_loss'])
loss_gap = best['train_loss'] - best['valid_loss']
print('best:', best)
print('train-valid loss gap:', round(loss_gap, 4))
```

### 해설 및 실행 결과 해석

- best epoch는 미리 정한 validation loss가 가장 낮은 시점입니다. 같은 epoch의 valid_loss - train_loss를 loss_gap으로 기록하면 accuracy gap과 혼동하지 않고 과적합 신호를 살필 수 있습니다.


## 문제 3. 리포트 문장 자동 생성하기

실험 결과 dictionary를 사용해 제출용 리포트 문장을 만듭니다.

```python
summary = {'best_model': 'CNN', 'valid_acc': 0.90, 'loss_gap': 0.11}
# TODO: summary 값을 사용해 report 문자열을 완성하세요.
report = (
    f'best model: {summary["best_model"]}\n'
    f'valid acc: {summary["valid_acc"]:.2f}\n'
    f'loss gap: {summary["loss_gap"]:.2f}'
)
print(report)
```

### 해설 및 실행 결과 해석

- 리포트는 숫자만 나열하는 것이 아니라 선택 기준, 결론, 한계까지 함께 적어야 합니다. loss_gap처럼 metric 종류를 이름에 명시하면 다른 gap과 혼동하지 않습니다.
