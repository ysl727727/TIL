```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. 문제 유형별 출력 형태 매핑
cases = [
    {'task': '집값 예측', 'type': '회귀'},
    {'task': '스팸 여부 판단', 'type': '이진 분류'},
    {'task': '10개 상품 카테고리 분류', 'type': '다중 분류'},
]

# TODO: output_dim, loss 후보를 채워보세요.
mapping = {
    '회귀': {'output_dim': 1, 'loss': 'MSELoss'},
    '이진 분류': {'output_dim': 1, 'loss': 'BCEWithLogitsLoss'},
    '다중 분류': {'output_dim': 10, 'loss': 'CrossEntropyLoss'},
}
for case in cases:
    print(case['task'], '->', mapping[case['type']])
```

```python
# 필수 2. 입력 데이터 형태 확인하기
tabular_x = torch.randn(5, 4)       # batch=5, feature=4
image_x = torch.randn(5, 1, 28, 28) # batch=5, channel=1, height=28, width=28
seq_x = torch.randn(5, 12, 8)       # batch=5, time=12, feature=8

# TODO: 각 텐서의 shape 설명을 작성해보세요.
shape_notes = {
    'tabular': '5개의 입력과 4개의 피쳐',
    'image': '5개의 이미지와 1개의 채널,높이 28 너비 28',
    'sequence': '5개의 입력과 12개의 시점, 8개의 피쳐'
}
print('tabular:', tabular_x.shape, shape_notes['tabular'])
print('image:', image_x.shape, shape_notes['image'])
print('sequence:', seq_x.shape, shape_notes['sequence'])
```

```python
# 심화 1. logits와 target을 보고 손실 함수 후보 판단하기
logits = torch.randn(6, 3)      # 3-class logits
target = torch.tensor([0, 2, 1, 1, 0, 2])

# TODO: 다중 분류에 맞는 loss_fn을 선택하세요.
loss_fn = nn.CrossEntropyLoss()
if loss_fn is None:
    print('TODO: nn.CrossEntropyLoss()를 연결해보세요.')
else:
    loss = loss_fn(logits, target)
    print('loss:', loss.item())
```
