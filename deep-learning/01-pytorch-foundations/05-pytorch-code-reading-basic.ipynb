# 1-5 Basic: Reading PyTorch Code Structure 긴 PyTorch 코드를 import, dataset, model, loss, optimizer, loop 단계로 나누고 forward, backward, update와 기록 위치를 확인했습니다. 상태: 기본 실습 완료. 학생 작성 셀과 실행 결과만 남겼습니다. 

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. 코드 블록에 단계명 붙이기
code_blocks = [
    'import torch, torch.nn as nn',
    'train_loader = ...',
    'model = nn.Linear(4, 3)',
    'loss_fn = nn.CrossEntropyLoss()',
    'optimizer = torch.optim.Adam(model.parameters())',
    'for epoch in range(epochs): ...',
]
stage_names = ['import', 'dataset', 'model', 'loss', 'optimizer', 'loop']
for block, stage in zip(code_blocks, stage_names):
    print(f'{block:<55} -> {stage}')
```

```python
# 필수 2. 작은 학습 코드 구조를 위에서 아래로 읽기
x = torch.randn(12, 4)
y = torch.randint(0, 3, (12,))
model = nn.Linear(4, 3)
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

logits = model(x) # forward
loss = loss_fn(logits, y) # loss
optimizer.zero_grad() ### gradient 초기화
loss.backward() # backward ### gradient 계산
optimizer.step() # update ### gradient 업데이트
print('logits shape:', logits.shape, '| loss:', round(loss.item(), 4))
```

```python
# 심화 1. metric과 checkpoint 위치 표시하기
history = {'loss': []}
checkpoint = {}

history['loss'].append(float(loss.item()))
checkpoint['model_state_dict'] = model.state_dict()
print('history:', history)
print('checkpoint keys:', list(checkpoint.keys()))
```
