# 2-2 Basic: Batch Dimension and Broadcasting `unsqueeze(0)`으로 batch 차원을 추가하고 broadcasting 결과 shape을 예측했습니다. Loss 전에 prediction과 target shape을 맞추는 이유도 확인했습니다.  상태: 기본 실습 완료. 학생 작성 셀과 실행 결과만 남겼습니다.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. batch dimension 추가/제거하기
single_image = torch.randn(1, 28, 28)  # C,H,W
batched_image = single_image.unsqueeze(0)  # 힌트: unsqueeze(0)
print('single_image shape:', single_image.shape)
print('batched_image shape:', batched_image.shape)
```

```python
# 필수 2. broadcasting 결과 shape 예측하기
a = torch.randn(4, 3)
b = torch.randn(3)
result = a + b
print('a shape:', a.shape)
print('b shape:', b.shape)
print('result shape:', result.shape)

# result.shape가 ([4,3])이 나오는 이유는 b의 앞축을 broadcasting으로 1이 채워지고,
# b의 앞축이 된 1은 4로 확장된다. 결과적으로 b의 행이 a의 행에 4번 더해진다.
```

```python
# 심화 1. 위험한 broadcasting 찾기
pred = torch.randn(4, 1)
target = torch.randn(4)  # shape이 다릅니다.
print('pred shape:', pred.shape, '| target shape:', target.shape)
fixed_target = target.unsqueeze(1)
print('fixed_target shape:', fixed_target.shape)

#### .view(4,1) 처럼 직접 구성할도 수 있음
```
