```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. 퍼셉트론 수식 계산하기
x = torch.tensor([2.0, -1.0])
w = torch.tensor([0.5, 1.0])
b = torch.tensor(0.2)
# TODO: z = x @ w + b 를 계산해보세요.
z = x @ w + b
print('z:', z)

assert torch.isclose(z, torch.tensor(0.2))
```

```python
# 필수 1. 퍼셉트론 수식 계산하기
x = torch.tensor([2.0, -1.0])
w = torch.tensor([0.5, 1.0])
b = torch.tensor(0.2)
# TODO: z = x @ w + b 를 계산해보세요.
z = x @ w + b
print('z:', z)

assert torch.isclose(z, torch.tensor(0.2))


# 필수 2. 선형 결정 경계 시각화 준비
points = torch.tensor([[1., 1.], [2., 1.], [-1., -1.], [-2., -1.]])
score = points @ w + b
pred = (score > 0).long()
print('pred:', pred)

```

```python
###

# 필수 2. 선형 결정 경계 시각화 준비
points = torch.tensor([[1., 1.], [2., 1.], [-1., -1.], [-2., -1.]])
scores = points @ w + b
pred = (scores > 0).long()
print('scores:', scores.tolist())
print('pred:', pred.tolist())
plt.scatter(points[:,0], points[:,1], c=pred)
plt.title('Perceptron predictions')
plt.xlabel('x1'); plt.ylabel('x2')
plt.show()
assert pred.shape == (4,)
```

```python
# 심화 1. bias를 바꾸면 결정 경계가 어떻게 이동하는지 관찰하기
bias_values = [-1.0, 0.0, 1.0]
for bias in bias_values:
    score = points @ w + bias
    print('bias=', bias, 'pred=', (score > 0).long().tolist())
```
