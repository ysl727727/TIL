# 3-3 가중치·편향과 nn.Linear

강의 원문 대신 직접 작성하고 실행한 코드와 학습 메모를 정리했습니다.


```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. nn.Linear의 weight/bias shape 확인
linear = nn.Linear(in_features=5, out_features=3)
print('weight shape:', linear.weight.shape)
print('bias shape:', linear.bias.shape)

# TODO: parameter 수를 직접 계산해보세요.
manual_param_count = 5 * 3 + 3
print('manual_param_count:', manual_param_count)

assert manual_param_count == sum(p.numel() for p in linear.parameters()) ## <<한번 더 보기
```

```python
# 필수 2. model.parameters()로 parameter 확인하기
for name, param in linear.named_parameters():
    print(name, param.shape, param.numel())
```

```python
###

# 필수 2. model.parameters()로 parameter 확인하기
for name, param in linear.named_parameters():
    print(name, param.shape, param.numel())
assert dict(linear.named_parameters())['weight'].shape == (3, 5)
```

```python
# 심화 1. 직접 계산한 Linear와 nn.Linear 비교하기
x = torch.randn(2, 5)
# TODO: y_manual = x @ W.T + b 형태를 완성해보세요.
y_layer = linear(x)
y_manual = x @ linear.weight.T + linear.bias
print('max difference:', (y_layer - y_manual).abs().max().item())

assert torch.allclose(y_layer, y_manual)
```
