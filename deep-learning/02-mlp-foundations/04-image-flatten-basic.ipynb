# 3-4 입출력 차원 계산과 flatten

강의 원문 대신 직접 작성하고 실행한 코드와 학습 메모를 정리했습니다.


```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. 이미지 Tensor flatten하기
images = torch.randn(8, 1, 28, 28)
# TODO: batch 차원을 유지하면서 flatten해보세요.
flat = torch.flatten(images, start_dim=1)
# 또는
flat2 = images.view(images.size(0), -1)
# 또는
flat3 = images.reshape(images.size(0), -1)
print('images shape:', images.shape)
print('flat shape:', flat.shape)
print('flat2 shape:', flat2.shape)
print('flat3 shape:', flat3.shape)
```

```python
# 필수 2. MLP 입력 차원 맞추기 *** 위 텍스트 설명이랑 헷갈리지 말기 image batch 사이즈 8임***
# TODO: flat의 feature 수에 맞게 in_features를 설정하세요.
in_features = flat.shape[1]
model = nn.Linear(in_features, 4)

out = model(flat)

print('model.in_features:', model.in_features)
print('out shape:', out.shape)

assert out.shape == (8, 4)
```

```python
# 심화 1. reshape/view 사용 시 batch 차원을 유지하기
small = torch.randn(2, 3, 4)
# TODO: small을 (2, 12)로 바꿔보세요.

small_flat = small.view(small.size(0), -1)
small_flat2 = small.view(small.size(0), -1)

print('small_flat shape:', small_flat.shape)
print('small_flat2 shape:', small_flat2.shape)

assert small_flat.shape and small_flat2.shape == (2, 12)
```
