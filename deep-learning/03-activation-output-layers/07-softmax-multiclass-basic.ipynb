# 4-4 Softmax와 다중 분류 출력층

강의 원문 대신 직접 작성하고 실행한 코드와 학습 메모를 정리했습니다.


```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)
```

```python
# 필수 1. logits를 Softmax 확률로 바꾸기
logits = torch.tensor([[2.0, 1.0, 0.1]])
# TODO: dim=1 기준 softmax를 적용해보세요.
probs = torch.softmax(logits, dim=1)
print('probs:', probs)
print('sum:', probs.sum())

assert torch.isclose(probs.sum(), torch.tensor(1.0))
```

```python
# 필수 2. argmax로 예측 class 만들기
logits = torch.tensor([[0.1, 2.5, 0.3], [3.0, 1.0, 0.2]])
# TODO: 각 sample의 예측 class index를 구하세요.
pred = torch.argmax(logits, dim=1)
print('pred:', pred.tolist())

assert (pred == torch.tensor([1, 0])).all() # pred를 .tolist or .all()로 모든원소비교 or torch.equal(x,y)사용
```

```python
# 심화 1. CrossEntropyLoss는 raw logits와 class index target을 받습니다.
logits = torch.randn(4, 3)
target = torch.tensor([0, 2, 1, 1], dtype=torch.long)
# TODO: loss를 계산해보세요.
loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(logits, target)
if loss_fn is None:
    print('TODO: nn.CrossEntropyLoss()를 사용해보세요.')
else:
    print(loss_fn(logits, target).item())

print(loss)
print(logits.shape)
print(target.dtype)

assert loss.ndim == 0
```
