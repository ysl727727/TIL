# 5-1 손실 함수의 역할과 목표 함수

강의 원문 대신 직접 작성한 코드, 실행 결과와 학습 메모를 정리했습니다.


```python
import random
import json
import math
import shutil
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader, Dataset, random_split

# 실습 결과가 매번 비슷하게 나오도록 seed를 고정합니다.
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('device:', device)
```

```python
pred = torch.tensor([[2.0], [4.0], [6.0]])
target = torch.tensor([[1.5], [3.0], [7.0]])
loss_fn = nn.MSELoss()

# TODO: 예측값과 정답의 차이를 제곱하세요.
squared_error = ((pred - target)**2)  # 임시 코드입니다. 제곱이 빠져 있습니다.

# TODO: squared_error의 평균을 구하세요.
manual_mse = squared_error.mean()

pytorch_mse = loss_fn(pred, target)
print('manual_mse:', float(manual_mse))
print('pytorch_mse:', float(pytorch_mse))

assert manual_mse == pytorch_mse
```

```python
target = torch.tensor([[1.0], [2.0], [3.0]])
candidates = torch.tensor([
    [[0.0], [0.0], [0.0]],
    [[0.8], [2.2], [3.1]],
    [[3.0], [2.0], [1.0]],
])

losses = []
for pred in candidates:
    # TODO: pred와 target 사이의 MSE를 계산해서 losses에 넣으세요.
    loss = ((pred - target)**2).mean()  # 임시 코드입니다.
    losses.append(float(loss))

# TODO: 가장 작은 loss의 위치를 구하세요.
best_idx = int(torch.tensor(losses).argmin())
print('losses:', losses)
print('best candidate index:', best_idx)
```

```python
binary_logits = torch.tensor([[0.2], [-1.0], [1.5]])
binary_target = torch.tensor([[1.0], [0.0], [1.0]])
multi_logits = torch.tensor([[2.0, 0.3, -0.5], [0.1, 1.4, 0.2]])
multi_target = torch.tensor([0, 1])

# TODO: 이진 분류 손실 함수를 만들고 loss를 계산하세요.
bce_loss = nn.BCEWithLogitsLoss()(binary_logits, binary_target)

# TODO: 다중 분류 손실 함수를 만들고 loss를 계산하세요.
ce_loss = nn.CrossEntropyLoss()(multi_logits, multi_target)

print('bce_loss:', bce_loss, 'ndim:', bce_loss.ndim)
print('ce_loss:', ce_loss, 'ndim:', ce_loss.ndim)
```
