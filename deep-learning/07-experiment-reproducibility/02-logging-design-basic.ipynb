## 필수 1 1: epoch log dict 만들기

### 문제 설명
epoch 결과를 dict 한 줄로 정리합니다.

```python
def make_epoch_log(epoch, train_loss, valid_loss, valid_acc):
    # TODO: epoch 로그 dict를 완성하세요.
    return {
        'epoch': epoch,
        'train_loss': train_loss,
        'valid_loss': valid_loss,
        'valid_acc': valid_acc
    }

print(make_epoch_log(2, 0.812345, 0.923456, 0.73456))
```

### 해설 및 실행 결과 해석
로그 dict는 나중에 DataFrame이나 JSON으로 저장하기 쉽습니다. 매 epoch 같은 key를 유지해야 비교와 시각화가 편합니다.

## 필수 2 2: 로그 리스트에 누적하기

### 문제 설명
여러 epoch 로그를 리스트에 누적합니다.

```python
logs = []
for epoch in range(1, 4):
    log = {'epoch': epoch, 'train_loss': 1/epoch, 'valid_loss': 1/(epoch+0.5), 'valid_acc': 0.5 + epoch*0.1}
    # TODO: logs에 log를 추가하세요.
    logs.append(log)
print(logs)
print('num_logs:', len(logs))
print('last:', logs[-1] if logs else None)
```

### 해설 및 실행 결과 해석
리스트에 dict를 쌓으면 epoch 단위 변화가 유지됩니다. 최종 리포트에서는 이 로그를 표나 그래프로 바꿔 사용합니다.
