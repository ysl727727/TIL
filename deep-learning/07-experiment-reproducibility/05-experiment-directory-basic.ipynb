```python
####
exp_dir = Path('runs/test_exp'); exp_dir.mkdir(parents=True, exist_ok=True)
config = {'lr': 0.01, 'batch_size': 16}
metrics = {'best_valid_loss': 0.42}
(exp_dir / 'config.json').write_text(json.dumps(config, ensure_ascii=False, indent=2), encoding='utf-8')
(exp_dir / 'metrics.json').write_text(json.dumps(metrics, ensure_ascii=False, indent=2), encoding='utf-8')
print('files:', sorted(p.name for p in exp_dir.iterdir()))
shutil.rmtree('runs', ignore_errors=True)
```

### 해설 및 실행 결과 해석
config와 metrics를 분리해두면 설정과 결과를 각각 쉽게 확인할 수 있습니다. 나중에 여러 실험을 비교할 때도 파일 구조가 일정해야 합니다.
